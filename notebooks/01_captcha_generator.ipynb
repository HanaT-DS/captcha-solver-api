```python
# Imports
import sys
sys.path.insert(0, '..')

import matplotlib.pyplot as plt
from app.services.captcha_generator import CaptchaGenerator, CharsetType, CHARSETS
```

```python
# Initialiser le générateur
generator = CaptchaGenerator()
print(f"Charset par défaut: {generator.charset}")
print(f"Taille du charset: {len(generator.charset)} caractères")
```

## 1. Génération de base

```python
# Générer un CAPTCHA
image, text = generator.generate()

plt.figure(figsize=(8, 3))
plt.imshow(image)
plt.title(f"CAPTCHA: {text}")
plt.axis('off')
plt.show()
```

## 2. Paramètres personnalisés

```python
# Différentes longueurs
fig, axes = plt.subplots(1, 4, figsize=(16, 3))

for ax, length in zip(axes, [3, 5, 7, 9]):
    image, text = generator.generate(length=length)
    ax.imshow(image)
    ax.set_title(f"Longueur {length}: {text}")
    ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Différents niveaux de bruit
fig, axes = plt.subplots(1, 4, figsize=(16, 3))

for ax, noise in zip(axes, [0.0, 0.3, 0.6, 0.9]):
    image, text = generator.generate(noise_level=noise)
    ax.imshow(image)
    ax.set_title(f"Bruit {noise}: {text}")
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## 3. Différents charsets

```python
# Charsets disponibles
for charset_type, charset in CHARSETS.items():
    print(f"{charset_type.value}: {charset[:30]}... ({len(charset)} chars)")
```

```python
# Générer avec différents charsets
fig, axes = plt.subplots(1, 3, figsize=(12, 3))

charsets = [
    ("CRNN (19 chars)", CHARSETS[CharsetType.CRNN]),
    ("Chiffres", CHARSETS[CharsetType.DIGITS]),
    ("Alphanumeric", CHARSETS[CharsetType.ALPHANUMERIC][:36]),
]

for ax, (name, charset) in zip(axes, charsets):
    gen = CaptchaGenerator(charset=charset)
    image, text = gen.generate()
    ax.imshow(image)
    ax.set_title(f"{name}: {text}")
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## 4. Génération pour les modèles

```python
# Génération optimisée pour CRNN (dimensions 300x75)
crnn_samples = generator.generate_for_model(model="crnn", n=3)

fig, axes = plt.subplots(1, 3, figsize=(15, 3))
for ax, (image, text) in zip(axes, crnn_samples):
    ax.imshow(image)
    ax.set_title(f"CRNN: {text} ({image.size})")
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## 5. Génération en batch

```python
# Générer 10 CAPTCHAs
batch = generator.generate_batch(n=10, noise_level=0.3)

fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for ax, (image, text) in zip(axes.flat, batch):
    ax.imshow(image)
    ax.set_title(text)
    ax.axis('off')

plt.tight_layout()
plt.show()
```
