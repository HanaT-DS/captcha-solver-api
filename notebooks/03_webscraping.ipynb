## 3. Exemple de scraping (simulation)

```python
# Exemple de configuration pour un site avec CAPTCHA
config = {
    "url": "https://example.com/login",
    "captcha_selector": "#captcha-image",
    "input_selector": "#captcha-input",
    "submit_selector": "#submit-btn",
    "model": "trocr",
}

print("📋 Configuration du scraping:")
for key, value in config.items():
    print(f"   {key}: {value}")
```

```python
# Fonction de scraping synchrone (pour démo)
async def demo_scrape():
    """Démonstration du scraping."""
    print("🌐 Initialisation du scraper...")
    
    scraper = ScraperService(default_model="trocr")
    
    # Prendre une screenshot d'un site public
    print("📸 Capture d'écran...")
    screenshot = await scraper.take_screenshot(
        url="https://httpbin.org/html",
        full_page=False,
    )
    
    # Sauvegarder
    with open("screenshot_demo.png", "wb") as f:
        f.write(screenshot)
    
    print("✅ Screenshot sauvegardé: screenshot_demo.png")
    
    await scraper.close()
    return screenshot

# Exécuter (décommenter pour tester)
# screenshot = await demo_scrape()
# display(Image(data=screenshot))
```
