```python
# Lister toutes les images .png
image_files = list(config.DATA_DIR.glob("*.png"))

print("Nombre total d'images :", len(image_files))

# Afficher quelques images aléatoires
sample_files = random.sample(image_files, 6)

plt.figure(figsize=(14, 4))
for i, f in enumerate(sample_files):
    img = plt.imread(f)
    label = os.path.splitext(os.path.basename(f))[0]  # label = nom du fichier
    plt.subplot(1, 6, i + 1)
    plt.imshow(img)
    plt.title(label)
    plt.axis("off")

plt.show()
```
