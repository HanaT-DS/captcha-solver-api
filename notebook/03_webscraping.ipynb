## 3. Test de base : Screenshot d'une page

```python
async def test_screenshot():
    scraper = ScraperService()
    
    # Prendre un screenshot
    screenshot = await scraper.take_screenshot(
        url="https://example.com",
        output_path="screenshot.png"
    )
    
    await scraper.close()
    
    print(f"Screenshot sauvegarde ({len(screenshot)} bytes)")
    return screenshot

# Executer
# screenshot = await test_screenshot()  # Dans Jupyter
# Ou en mode synchrone:
# screenshot = asyncio.run(test_screenshot())
```

## 6. Selecteurs CSS courants

Le scraper detecte automatiquement les CAPTCHAs avec ces selecteurs :

### Images CAPTCHA
- `img[src*='captcha']`
- `img[id*='captcha']`
- `#captcha`
- `.captcha`

### Champs input
- `input[name*='captcha']`
- `input[id*='captcha']`
- `#captcha-input`
