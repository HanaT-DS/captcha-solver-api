## 1. Import du module

```python
import sys
from pathlib import Path

# Ajouter le repertoire parent au path
sys.path.insert(0, str(Path().resolve().parent))

from app.services.captcha_generator import CaptchaGenerator
import matplotlib.pyplot as plt

print("Import reussi")
```

---
## 2. Generation d'un CAPTCHA simple

```python
# Creer le generateur
generator = CaptchaGenerator()

# Generer un CAPTCHA
image, text = generator.generate()

print(f"Texte genere : {text}")
print(f"Dimensions : {image.size}")

# Afficher
plt.figure(figsize=(8, 3))
plt.imshow(image)
plt.title(f"CAPTCHA : {text}")
plt.axis("off")
plt.show()
```

---
## 3. Variation du niveau de bruit

```python
noise_levels = [0.0, 0.3, 0.5, 0.8]

fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for ax, noise in zip(axes, noise_levels):
    image, text = generator.generate(noise_level=noise)
    ax.imshow(image)
    ax.set_title(f"Bruit: {noise}\nTexte: {text}")
    ax.axis("off")

plt.suptitle("Effet du niveau de bruit", fontsize=14)
plt.tight_layout()
plt.show()
```

---
## 4. Generation en lot

```python
# Generer 10 CAPTCHAs
batch = generator.generate_batch(n=10)

print(f"Nombre genere : {len(batch)}")

# Afficher
fig, axes = plt.subplots(2, 5, figsize=(15, 6))

for ax, (image, text) in zip(axes.flatten(), batch):
    ax.imshow(image)
    ax.set_title(text)
    ax.axis("off")

plt.suptitle("Lot de 10 CAPTCHAs", fontsize=14)
plt.tight_layout()
plt.show()
```
